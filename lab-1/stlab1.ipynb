```python
import numpy as np
import matplotlib.pyplot as plt

def true_probability(r, a):
    """Вычисление истинной геометрической вероятности как отношение площадей."""
    return (np.pi * r**2) / (4 * a**2)

def monte_carlo_simulation(r, a, n):
    """Генерация n случайных точек и определение их попадания в круг радиуса r."""
    x = np.random.uniform(-a, a, n)
    y = np.random.uniform(-a, a, n)
    inside_circle = (x**2 + y**2) <= r**2
    return np.sum(inside_circle) / n

def estimate_probability(r, a, max_n):
    """Оценка вероятности попадания точек в круг при увеличении n."""
    true_p = true_probability(r, a)
    ns = np.logspace(1, np.log10(max_n), num=50, dtype=int)
    estimates = np.array([monte_carlo_simulation(r, a, n) for n in ns])
    errors = np.abs(estimates - true_p)
    return ns, estimates, errors

def required_n_for_epsilon(r, a, epsilons, max_n=100000):
    """Определение N(ε) для различных значений точности."""
    ns, estimates, errors = estimate_probability(r, a, max_n)
    required_n = [ns[np.where(errors < eps)[0][0]] if np.any(errors < eps) else max_n for eps in epsilons]
    return epsilons, required_n

# Параметры
a = 1  # Половина стороны квадрата
r_values = [a / (k + 1) for k in range(1, 6)]  # 5 значений радиуса
max_n = 100000  # Максимальное число точек

epsilons = [1e-1, 1e-2, 1e-3, 1e-4]
```

```python
plt.figure(figsize=(12, 6))

for r in r_values:
    ns, estimates, errors = estimate_probability(r, a, max_n)
    plt.plot(ns, estimates, label=f"r = {r:.2f}")

plt.xscale("log")
plt.xlabel("Число точек n")
plt.ylabel("Оценка вероятности p")
plt.legend()
plt.grid()
plt.title("Сходимость оценки вероятности p(n)")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
for r in r_values:
    ns, estimates, errors = estimate_probability(r, a, max_n)
    plt.plot(ns, errors, label=f"r = {r:.2f}")

plt.xscale("log")
plt.yscale("log")
plt.xlabel("Число точек n")
plt.ylabel("Ошибка оценки ε(n)")
plt.legend()
plt.grid()
plt.title("Изменение ошибки оценки вероятности ε(n)")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
for r in r_values:
    epsilons, required_ns = required_n_for_epsilon(r, a, epsilons, max_n)
    plt.plot(epsilons, required_ns, marker='o', label=f"r = {r:.2f}")

plt.xscale("log")
plt.yscale("log")
plt.xlabel("Точность ε")
plt.ylabel("Необходимое число точек N(ε)")
plt.legend()
plt.grid()
plt.title("График N(ε)")
plt.show()
```
